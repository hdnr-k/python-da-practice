```python
# 32.How to compute the autocorrelations of a numeric series?
# Compute autocorrelations for the first 10 lags of ser. Find out which lag has the largest correlation.
np.random.seed(2)
ser = pd.Series(np.arange(20) + np.random.normal(1, 10, 20))
ser.head()
```

```python
autocorr_list = [ser.autocorr(lag=lag) for lag in range(1, 11)]
autocorr_list
```

```python
autocorr_series = pd.Series(autocorr_list, index=range(1, 11))
autocorr_series
```

```python
largest_lag = autocorr_series.abs().idxmax()
largest_lag
# abs() 数字のプラス・マイナスを取り除いて、絶対値にする
# idxmax() 一番大きい値を持つ要素のインデックスを取得する
```

```python
# 32.Solution
autocorrelations = [ser.autocorr(i).round(2) for i in range(11)]
autocorrelations[1:]
print('Lag having highest correlation: ', np.argmax(np.abs(autocorrelations[1:]))+1)
```

```python
# 35.How to create a dataframe with rows as strides from a given series?
L = pd.Series(range(15))
```

```python
window_size = 5
step_size = 2
```

```python
range(0, len(L) - window_size + 1, step_size)
```

```python
stride_list = [
    L.iloc[i: i + window_size].values
    for i in range(0, len(L) - window_size +1, step_size)
]
stride_list
```

```python
pd.DataFrame(stride_list)
```

```python
# 35.Solution
def gen_strides(a, stride_len=5, window_len=5):
 n_strides = ((a.size-window_len)//stride_len) + 1
 return np.array([a[s:(s+window_len)] for s in np.arange(0, a.size, stride_len)[:n_strides]])

gen_strides(L, stride_len=2, window_len=4)
```
